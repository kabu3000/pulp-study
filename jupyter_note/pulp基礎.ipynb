ある工場で、製品Aと製品Bを作っています。
​製品A：1個作ると 3万円 の利益。作るのに原料が 2kg 必要。
​製品B：1個作ると 5万円 の利益。作るのに原料が 1kg 必要。
​制約：使える原料は全部で 10kg まで。
​目標：利益を最大にするには、それぞれ何個作ればよいか？

```python
import pulp

# 1. 問題の定義（LpMaximize: 利益を最大化したい）
prob = pulp.LpProblem("Production_Plan", pulp.LpMaximize)

# 2. 変数の定義（lowBound=0: 作る個数は0以上）
x1 = pulp.LpVariable("Product_A", lowBound=0)
x2 = pulp.LpVariable("Product_B", lowBound=0)

# 3. 目的関数（最大化したい利益の式）
prob += 3 * x1 + 5 * x2, "Total_Profit"

# 4. 制約条件（原料は10kgまで）
prob += 2 * x1 + 1 * x2 <= 10, "Material_Constraint"

# 5. 解く
status = prob.solve()

# 結果表示
print(f"ステータス: {pulp.LpStatus[status]}")
print(f"製品Aの生産数: {x1.varValue}")
print(f"製品Bの生産数: {x2.varValue}")
print(f"最大利益: {pulp.value(prob.objective)} 万円")
```

```python
#2日目

import pulp

# データ定義
factories = ["Kumamoto", "Sendai"]
stores = ["Tokyo", "Osaka", "Fukuoka"]

supply = {"Kumamoto": 50, "Sendai": 60}
demand = {"Tokyo": 40, "Osaka": 40, "Fukuoka": 20}

costs = {
    ("Kumamoto", "Tokyo"): 4, ("Kumamoto", "Osaka"): 2, ("Kumamoto", "Fukuoka"): 1,
    ("Sendai", "Tokyo"): 1, ("Sendai", "Osaka"): 3, ("Sendai", "Fukuoka"): 5,
}

# 問題定義
prob = pulp.LpProblem("Trans_Cost_Min", pulp.LpMinimize)

# 変数定義（工場と店舗の全組み合わせの変数を作る）
# route_vars[f, s] は「工場fから店舗sへ運ぶ量」
routes = [(f, s) for f in factories for s in stores]
x = pulp.LpVariable.dicts("Route", routes, lowBound=0, cat='Integer')

#routesはcostsで定義されている拠点間名である。x(f,s)が輸送量となる
#二日目
```

```python
# --- ここから下を自分で考えてみてください ---

import shutil

# 目的関数：輸送コストの合計を最小化
# ヒント：pulp.lpSum([costs[f, s] * x[f, s] for f, s in routes])

# 制約条件1：各工場の出荷量は、供給能力以下
# ヒント：工場ごとに「出荷量の合計 <= supply[f]」

# 制約条件2：各店舗の受取量は、需要量以上（またはピッタリ）
# ヒント：店舗ごとに「受取量の合計 >= demand[s]」



# 輸送コストの合計を最小化
prob += pulp.lpSum([costs[f, s] * x[f, s] for f, s in routes]), "Total_Cost"

#供給制限 熊本からの供給は50個まで
for f in factories:
    prob += pulp.lpSum([x[f, s] for s in stores]) <= supply[f], f"Supply_{f}"
#需要制限
for s in stores:
        prob += pulp.lpSum([x[f, s] for f in factories]) == demand[s], f"Demand_{s}"

# 熊本から大阪へのルートは15個が限界！
prob += x[("Kumamoto", "Osaka")] <= 15


# 6. 解く
glpsol_path = shutil.which("glpsol")
if glpsol_path:
    solver = pulp.GLPK_CMD(path=glpsol_path, msg=False)
else:
    solver = pulp.PULP_CBC_CMD(msg=False)
prob.solve(solver)

```

```python
# 6. 結果の表示
print(f"ステータス: {pulp.LpStatus[prob.status]}")
print(f"総輸送コスト: {pulp.value(prob.objective)} 円")
print("-" * 30)

# 各ルートの輸送量を表示
for f in factories:
    for s in stores:
        amount = x[f, s].varValue
        if amount > 0: # 運ぶ量が0より大きいルートだけ表示
            print(f"【輸送】 {f} 工場 ➔ {s} 店舗 : {amount} 個 (単価: {costs[f,s]}円)")
```

```python
#可能な限り均一化された勤務日数での最適化実施
import pulp
import pandas as pd

# 1. 問題の定義（LpMinimize = ペナルティを最小化する）
prob = pulp.LpProblem("Shift_Scheduling_Goal_Programming", pulp.LpMinimize)

# 2. リストの定義
staffs = ["A", "B", "C", "D", "E", "F", "G"]
days = ["月", "火", "水", "木", "金", "土", "日"]
slots = ["朝", "昼", "夜", "深夜"]

# 3. 各曜日・各時間帯に必要な人数（需要：合計42枠）
required = {}
for d in days:
    for s in slots:
        if s == "朝": required[(d, s)] = 1
        elif s == "昼": required[(d, s)] = 2
        elif s == "夜": required[(d, s)] = 2
        elif s == "深夜": required[(d, s)] = 1

# 4. 変数の定義
# シフトの変数（0か1）
x = pulp.LpVariable.dicts("Shift", 
                          ((d, f, s) for d in days for f in staffs for s in slots), 
                          cat=pulp.LpBinary)

# 🌟 プランBの秘密兵器：「理想（6日）からのズレ」を表す変数
# 連続値（Continuous）として、0以上の値（lowBound=0）をとるようにします
penalty_plus = pulp.LpVariable.dicts("PenaltyPlus", staffs, lowBound=0, cat=pulp.LpContinuous)   # 理想より多い分
penalty_minus = pulp.LpVariable.dicts("PenaltyMinus", staffs, lowBound=0, cat=pulp.LpContinuous) # 理想より少ない分


# 🌟 --- 目的関数の設定（ここがプランBの主役） --- 🌟
# 「全員のズレ（ペナルティ）の合計」を一番小さくすることを目標にします
prob += pulp.lpSum([penalty_plus[p] + penalty_minus[p] for p in staffs])


# --- 5. 制約条件 ---

# ルール①：各時間帯の必要人数をピッタリ満たす
for d in days:
    for s in slots:
        prob += pulp.lpSum([x[d, p, s] for p in staffs]) == required[(d, s)]

# ルール②：かならず「1日1シフト」に限定する
for d in days:
    for p in staffs:
        prob += pulp.lpSum([x[d, p, s] for s in slots]) <= 1

# ルール③：深夜勤務の後は、翌日の「朝・昼・夜」は休暇にする
for i in range(len(days) - 1):
    today = days[i]
    tomorrow = days[i+1]
    for p in staffs:
        prob += x[today, p, "深夜"] + pulp.lpSum([x[tomorrow, p, s] for s in ["朝", "昼", "夜"]]) <= 1

# 🌟 プランBの制約：出勤数と「ズレ（ペナルティ）」を連動させる
# 「実際の出勤数」−「理想より多い分」＋「理想より少ない分」＝ 6日（理想）
# という方程式を組むことで、ソルバーにズレを自動計算させます
ideal_days = 6
for p in staffs:
    actual_work_days = pulp.lpSum([x[d, p, s] for d in days for s in slots])
    prob += actual_work_days - penalty_plus[p] + penalty_minus[p] == ideal_days


# --- 6. ソルバー実行 ---
status = prob.solve()

# --- 7. 結果の表示とCSV出力 ---
print(f"ステータス: {pulp.LpStatus[status]}")
print("-" * 30)

if status == 1:
    row_indices = [(d, s) for d in days for s in slots]
    df_index = pd.MultiIndex.from_tuples(row_indices, names=["曜日", "シフト"])
    df = pd.DataFrame(index=df_index, columns=staffs, data="")
    
    for d in days:
        for s in slots:
            for p in staffs:
                if pulp.value(x[d, p, s]) == 1:
                    df.loc[(d, s), p] = "○"
    
    # 画面表示
    print(df)
    
    # 集計結果の確認
    print("\n▼ 各スタッフの合計出勤日数（プランBの結果）")
    for p in staffs:
        total = sum(1 for d in days for s in slots if pulp.value(x[d, p, s]) == 1)
        print(f"  {p}さん: {total}日 (理想の6日から {pulp.value(penalty_plus[p]) - pulp.value(penalty_minus[p])} のズレ)")
        
    df.to_csv("shift_table_plan_b.csv", encoding="utf_8_sig")
    print("-" * 30)
    print("🎉 プランBで最適化された『 shift_table_plan_b.csv 』を出力しました！")
else:
    print("⚠️ パズルを解くことができませんでした（Infeasible）")
```
