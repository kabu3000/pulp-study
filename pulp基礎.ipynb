ある工場で、製品Aと製品Bを作っています。
​製品A：1個作ると 3万円 の利益。作るのに原料が 2kg 必要。
​製品B：1個作ると 5万円 の利益。作るのに原料が 1kg 必要。
​制約：使える原料は全部で 10kg まで。
​目標：利益を最大にするには、それぞれ何個作ればよいか？

```python
import pulp

# 1. 問題の定義（LpMaximize: 利益を最大化したい）
prob = pulp.LpProblem("Production_Plan", pulp.LpMaximize)

# 2. 変数の定義（lowBound=0: 作る個数は0以上）
x1 = pulp.LpVariable("Product_A", lowBound=0)
x2 = pulp.LpVariable("Product_B", lowBound=0)

# 3. 目的関数（最大化したい利益の式）
prob += 3 * x1 + 5 * x2, "Total_Profit"

# 4. 制約条件（原料は10kgまで）
prob += 2 * x1 + 1 * x2 <= 10, "Material_Constraint"

# 5. 解く
status = prob.solve()

# 結果表示
print(f"ステータス: {pulp.LpStatus[status]}")
print(f"製品Aの生産数: {x1.varValue}")
print(f"製品Bの生産数: {x2.varValue}")
print(f"最大利益: {pulp.value(prob.objective)} 万円")
```

そのため 製品Bに全振りする回答を出した。<br>
ここで追加条件を出す<br>
- 作業時間: 製品Aは 3時間、製品Bは 1時間 かかる。
- 時間の制約: 合計で 15時間 までしか働けない。
- 結果として製品B 全振りの状況は変わらない

```python
import pulp

# 1. 問題の定義（LpMaximize: 利益を最大化したい）
prob = pulp.LpProblem("Production_Plan", pulp.LpMaximize)

# 2. 変数の定義（lowBound=0: 作る個数は0以上）
x1 = pulp.LpVariable("Product_A", lowBound=0)
x2 = pulp.LpVariable("Product_B", lowBound=0)

# 3. 目的関数（最大化したい利益の式）
prob += 3 * x1 + 5 * x2, "Total_Profit"

# 4. 制約条件（原料は10kgまで）
prob += 2 * x1 + 1 * x2 <= 10, "Material_Constraint"
prob += 3 *x1 + 1 *x2 <=15 , "Time_Constraiant" #製造時間の上限を追加」 

# 5. 解く
status = prob.solve()

# 結果表示
print(f"ステータス: {pulp.LpStatus[status]}")
print(f"製品Aの生産数: {x1.varValue}")
print(f"製品Bの生産数: {x2.varValue}")
print(f"最大利益: {pulp.value(prob.objective)} 万円")
```

```python
import pulp

# 1. データの定義（ここを変えるだけで条件変更が楽になります）
products = ["A", "B"]
profits = {"A": 3, "B": 5}
materials = {"A": 2, "B": 1}
hours = {"A": 3, "B": 1}

# 2. 問題の定義
prob = pulp.LpProblem("Smart_Production_Plan", pulp.LpMaximize)

# 3. 変数の定義（辞書形式で一括作成）
x = pulp.LpVariable.dicts("Prod", products, lowBound=0)

# 4. 目的関数（lpSumを使って合計を出す）
prob += pulp.lpSum([profits[i] * x[i] for i in products]), "Total_Profit"

# 5. 制約条件
prob += pulp.lpSum([materials[i] * x[i] for i in products]) <= 10, "Material"
prob += pulp.lpSum([hours[i] * x[i] for i in products]) <= 15, "Time"

# 個別制約（Aは2個以上、Bは6個以下）
prob += x["A"] >= 2
prob += x["B"] <= 6

# 6. 解く
prob.solve()

for i in products:
    print(f"製品{i}の生産数: {x[i].varValue}")
```

```python
#2日目

import pulp

# データ定義
factories = ["Kumamoto", "Sendai"]
stores = ["Tokyo", "Osaka", "Fukuoka"]

supply = {"Kumamoto": 50, "Sendai": 60}
demand = {"Tokyo": 40, "Osaka": 40, "Fukuoka": 20}

costs = {
    ("Kumamoto", "Tokyo"): 4, ("Kumamoto", "Osaka"): 2, ("Kumamoto", "Fukuoka"): 1,
    ("Sendai", "Tokyo"): 1, ("Sendai", "Osaka"): 3, ("Sendai", "Fukuoka"): 5,
}

# 問題定義
prob = pulp.LpProblem("Trans_Cost_Min", pulp.LpMinimize)

# 変数定義（工場と店舗の全組み合わせの変数を作る）
# route_vars[f, s] は「工場fから店舗sへ運ぶ量」
routes = [(f, s) for f in factories for s in stores]
x = pulp.LpVariable.dicts("Route", routes, lowBound=0, cat='Integer')

#routesはcostsで定義されている拠点間名である。x(f,s)が輸送量となる
#二日目
```

```python
# --- ここから下を自分で考えてみてください ---

# 目的関数：輸送コストの合計を最小化
# ヒント：pulp.lpSum([costs[f, s] * x[f, s] for f, s in routes])

# 制約条件1：各工場の出荷量は、供給能力以下
# ヒント：工場ごとに「出荷量の合計 <= supply[f]」

# 制約条件2：各店舗の受取量は、需要量以上（またはピッタリ）
# ヒント：店舗ごとに「受取量の合計 >= demand[s]」



# 輸送コストの合計を最小化
prob += pulp.lpSum([costs[f, s] * x[f, s] for f, s in routes]), "Total_Cost"

#供給制限 熊本からの供給は50個まで
for f in factories:
    prob += pulp.lpSum([x[f, s] for s in stores]) <= supply[f], f"Supply_{f}"
#需要制限
for s in stores:
        prob += pulp.lpSum([x[f, s] for f in factories]) == demand[s], f"Demand_{s}"

# 熊本から大阪へのルートは15個が限界！
prob += x[("Kumamoto", "Osaka")] <= 15


# 6. 解く
solver = pulp.GLPK_CMD(path="/data/data/com.termux/files/usr/bin/glpsol")
prob.solve(solver)

```

```python
# 6. 結果の表示
print(f"ステータス: {pulp.LpStatus[prob.status]}")
print(f"総輸送コスト: {pulp.value(prob.objective)} 円")
print("-" * 30)

# 各ルートの輸送量を表示
for f in factories:
    for s in stores:
        amount = x[f, s].varValue
        if amount > 0: # 運ぶ量が0より大きいルートだけ表示
            print(f"【輸送】 {f} 工場 ➔ {s} 店舗 : {amount} 個 (単価: {costs[f,s]}円)")
```
